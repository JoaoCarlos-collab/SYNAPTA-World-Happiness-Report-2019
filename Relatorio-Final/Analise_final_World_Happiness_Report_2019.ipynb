```python
!pip install pycountry_convert
import pycountry_convert as pc

# 1. Função para traduzir o nome do país e GARANTIR a criação da coluna 'Continente'
def obter_continente(nome_pais):
    try:
        if nome_pais == 'United States': return 'América do Norte'
        if nome_pais == 'United Kingdom': return 'Europa'
        if nome_pais == 'South Korea': return 'Ásia'

        codigo_pais = pc.country_name_to_country_alpha2(nome_pais)
        codigo_continente = pc.country_alpha2_to_continent_code(codigo_pais)

        nomes_continentes = {
            'NA': 'América do Norte', 'SA': 'América do Sul',
            'AS': 'Ásia', 'OC': 'Oceânia', 'AF': 'África', 'EU': 'Europa'
        }
        return nomes_continentes.get(codigo_continente, 'Outro')
    except:
        return 'Outro'

# Aplica a função no df atual para criar a coluna antes dos cálculos
df['Continente'] = df['Country or region'].apply(obter_continente)

# 2. Calculando a Média e Mediana GLOBAL (Mundo todo)
media_global = df['Score'].mean()
mediana_global = df['Score'].median()

print(f"🌍 ESTATÍSTICAS GLOBAIS DE FELICIDADE:")
print(f"Média Global: {media_global:.2f}")
print(f"Mediana Global: {mediana_global:.2f}")
print("-" * 40)

# 3. Calculando a Média e Mediana por CONTINENTE
# Agora a coluna 'Continente' existe na memória e o agrupamento funcionará
estatisticas_continente = df.groupby('Continente')['Score'].agg(['mean', 'median']).reset_index()

# Renomeando as colunas para ficar em português e mais apresentável
estatisticas_continente.rename(columns={'mean': 'Média', 'median': 'Mediana'}, inplace=True)

# Ordenando pela Média (do mais feliz para o menos)
estatisticas_continente = estatisticas_continente.sort_values(by='Média', ascending=False)

print(f"\n📊 ESTATÍSTICAS DE FELICIDADE POR CONTINENTE:")
display(estatisticas_continente)
```

A média revela a capacidade total ou o peso matemático da região. Ela mostra qual seria a nota do continente se a felicidade ou a riqueza fossem distribuídas de forma perfeitamente igual entre todos os países.

A mediana revela o cenário exato e mais comum. Ela mostra a nota do país que está exatamente no centro da distribuição, permitindo concluir qual é o verdadeiro padrão de vida enfrentado pela maior parte da região.

Uma média igual à mediana reflete uma distribuição perfeitamente equilibrada onde quase todos vivem a mesma realidade.
Uma média maior que a mediana indica que uma pequena minoria privilegiada puxou o cálculo para cima, mascarando a desvantagem da grande maioria.
Por outro lado, uma média menor que a mediana mostra que a regra geral da região é de prosperidade, mas uma minoria isolada em extrema desvantagem conseguiu derrubar a nota matemática. O seu raciocínio interpretativo está absolutamente correto.

A Média (7.267500) e a Mediana (7.2675) apresentam valores rigorosamente iguais. Existe uma igualdade perfeita. A qualidade de vida e a felicidade são distribuídas de forma homogénea entre os países que compõem o grupo, sem distorções ou abismos sociais.

A Média (6.278513) é superior à Mediana (6.1820). Observa-se uma desigualdade mascarada. Uma elite de países com pontuações altíssimas puxa o cálculo matemático para cima, escondendo o facto de que a verdadeira realidade da maioria dos países europeus é inferior à apresentada pela média.

A Média (6.151583) é inferior à Mediana (6.2870). Regista-se uma crise isolada. O padrão da maioria da região é superior ao que a média indica, revelando que a nota foi arrastada para baixo por um número reduzido de países com piores condições.

A Média (5.920200) é inferior à Mediana (6.0570). Partilha do mesmo cenário da América do Norte, evidenciando uma crise isolada. A regra geral sul-americana apresenta um padrão melhor e mais feliz, mas nações específicas em forte desvantagem derrubam o resultado matemático do continente.   

Ásia:

A esmagadora maioria dos países asiáticos partilha da mesma realidade estatística, não havendo elites proeminentes ou falhas isoladas a distorcer os dados.   

Outro:
A Média (5.243600) é significativamente superior à Mediana (4.8120). Este é o caso mais flagrante de falsa prosperidade no dataset. Um grupo extremamente reduzido com notas elevadas inflaciona a matemática, ocultando a dura realidade enfrentada pela esmagadora maioria, que se encontra num patamar de bem-estar bastante inferior.   

África:
A Média (4.356814) é inferior à Mediana (4.4610). Configura um cenário de crise extrema isolada.
```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Separar os 10 mais felizes e contar os continentes
top_10 = df.nlargest(10, 'Score')
contagem_top_10 = top_10['Continente'].value_counts().reset_index()
contagem_top_10.columns = ['Continente', 'Quantidade de Países']

# 2. Separar os 10 menos felizes e contar os continentes
bottom_10 = df.nsmallest(10, 'Score')
contagem_bottom_10 = bottom_10['Continente'].value_counts().reset_index()
contagem_bottom_10.columns = ['Continente', 'Quantidade de Países']

# 3. Criar a figura com dois gráficos lado a lado (1 linha, 2 colunas)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Top 10
sns.barplot(
    data=contagem_top_10, x='Continente', y='Quantidade de Países',
    ax=axes[0], hue='Continente', palette='Blues_r', legend=False
)
axes[0].set_title('Continentes dos 10 Países MAIS Felizes', fontsize=12, fontweight='bold')
axes[0].set_ylim(0, 10) # Trava o eixo Y em 10 para os dois gráficos ficarem na mesma escala

# Gráfico 2: Bottom 10
sns.barplot(
    data=contagem_bottom_10, x='Continente', y='Quantidade de Países',
    ax=axes[1], hue='Continente', palette='Reds_r', legend=False
)
axes[1].set_title('Continentes dos 10 Países MENOS Felizes', fontsize=12, fontweight='bold')
axes[1].set_ylim(0, 10)

# Ajustar o layout para não cortar textos
plt.tight_layout()
plt.show()
```

Isso mostra que governos transparentes servem como base para melhorar os outros fatores que trazem felicidade às pessoas.

2. A Falta de Confiança Diminui a Ajuda ao Próximo:

Quando as pessoas sentem que têm pouca Freedom To Make Life Choices (que mede
o nível de liberdade individual) e percebem muita corrupção, ou seja, notas ruins em
Perceptions Of Corruption (que mede o nível de corrupção percebida e a confiança no
governo), elas tendem a ajudar menos os outros. Os dados mostram que regiões com
os piores resultados nesses dois pontos (como América do Sul, África, Europa e a
categoria "Outros") também têm as piores notas em Generosity (que reflete o quão
disposta a população está a fazer doações e ajudar o próximo). Isso indica que a
generosidade não é apenas uma questão de cultura, mas também de confiança. Se a
população sofre com a corrupção e com a falta de liberdade, a confiança na sociedade
diminui, o que reduz as doações e o apoio ao próximo.

3. A Importância do Apoio Social para a Saúde:

Os dados mostram que a falta de um bom Social Support (que representa a rede de
amparo formada por familiares, amigos ou assistência do Estado) está diretamente
ligada a uma pior Healthy Life Expectancy (que indica a expectativa de vida com
saúde de qualidade). A África, a região "Outros" e a Ásia têm as piores notas nesses
dois indicadores. Isso prova que viver muito e com qualidade não depende apenas de
hospitais, mas de ter pessoas com quem contar. Sem uma boa rede de apoio, as
pessoas ficam mais vulneráveis, o que prejudica a sua saúde física e mental ao longo
dos anos.

4. O Valor da Liberdade e do Apoio:

A análise mostra que ter um alto Gdp Per Capita (que representa a riqueza econômica e o Produto Interno Bruto por pessoa) não garante, por si só, que a população será feliz. Um exemplo claro é a diferença entre a Ásia e a América do Norte. Embora a Ásia possua o terceiro maior PIB do estudo, a região possui países listados no Top 10 dos menos felizes do mundo e figura como o terceiro continente com a menor média geral de felicidade. Por outro lado, a América do Norte possui o terceiro pior PIB nesta comparação, mas destaca-se por colocar os seus países no Top 10 dos mais felizes do mundo, sendo o terceiro continente com a maior média de felicidade. O motivo desta forte contradição? A América do Norte tem notas muito melhores em Healthy Life Expectancy (que indica a expectativa de vida com saúde de qualidade), Social Support (que representa a rede de amparo formada por familiares, amigos ou assistência do Estado) e Freedom To Make Life Choices (que mede o nível de liberdade individual). Isso conclui que o dinheiro só traz felicidade quando as pessoas também têm saúde, apoio e liberdade para tomar as suas próprias decisões.

5. A Generosidade Nascida da Necessidade:

A situação da Ásia nos mostra algo muito interessante. Apesar de ter as piores notas em
Social Support (que representa a rede de amparo formada por familiares, amigos ou
assistência do Estado), Freedom To Make Life Choices (que mede o nível de liberdade
individual) e Healthy Life Expectancy (que indica a expectativa de vida com saúde de
qualidade), a região tem a segunda melhor nota do mundo em Generosity (que reflete o
quão disposta a população está a fazer doações e ajudar o próximo). No começo,
podemos pensar que isso acontece apenas porque as pessoas, ao passarem por
dificuldades, sentem mais empatia umas pelas outras. Porém, olhando mais a fundo,
vemos que a própria falta de Freedom To Make Life Choices e de apoio
governamental cria essa situação. Como as pessoas não têm ajuda oficial e possuem
pouca liberdade para mudar de vida sozinhas, a própria cultura as obriga a se ajudarem
para sobreviver. Os dados mostram a formação de uma rede de apoio criada pela
própria sociedade, motivada tanto pela empatia quanto pela necessidade e obrigação de
cuidar da comunidade.

Correlação e Causalidade:

É importante destacar que as associações matemáticas mostradas neste relatório indicam
tendências, e não uma relação direta de causa e efeito. O fato de dois indicadores crescerem juntos
— como um forte Social Support acompanhado de uma longa expectativa de vida — não prova que
um gera o outro, pois elementos externos como a cultura e a história local também influenciam
diretamente a sociedade. Assim, a análise revela padrões que costumam ocorrer ao mesmo tempo,
mas não define regras absolutas de que um fator isolado seja o único responsável pela felicidade de
um país.

Principais Limitações da Análise

Apesar das descobertas relevantes, é fundamental reconhecer duas limitações centrais neste estudo. A primeira diz respeito à temporalidade: como o conjunto de dados é de 2019, torna-se um desafio aplicar as suas conclusões de forma direta aos dias de hoje. O cenário global passou por transformações sociais e econômicas profundas desde então, fazendo com que estes dados representem um retrato de uma época, e não a realidade atual.

A segunda limitação reside na restrição das variáveis. O dataset é bastante resumido e de natureza estritamente quantitativa, carecendo de informações fundamentais como as particularidades culturais, o histórico sociopolítico e a estrutura da sociedade de cada país.

Devido a essa ausência de contexto e à natureza altamente condensada dos dados, não é viável extrair conclusões definitivas baseando-se exclusivamente nestes números. O relatório atua como um excelente ponto de partida matemático, mas exige que a análise seja complementada com pesquisas externas e estudos qualitativos para compreender a verdadeira complexidade da felicidade nas diferentes nações.

1. Os Fatores Associados às Maiores Pontuações:

Como provado na análise detalhada dos dados, o dinheiro (representado pelo GDP per capita) sozinho não dita o topo do ranking global. O que realmente impulsiona os países para as maiores pontuações de felicidade é a combinação de um forte Social Support (ter com quem contar) e uma alta Healthy Life Expectancy (viver muito e com saúde de qualidade), aliados também à Freedom to Make Life Choices (liberdade para fazer escolhas e guiar o próprio caminho).
Estes fatores demonstraram ser os pilares mais fortemente associados ao bem-estar da população, superando, muitas vezes, o impacto isolado da riqueza financeira.

2. Cuidados na Interpretação e Limitações:

No entanto, o principal cuidado ao interpretar esta relação é lembrar que os dados mostram apenas uma correlação estatística, e não uma relação de causa e efeito direta.
Como o dataset utilizado é extremamente focado em números absolutos e métricas quantitativas, ele não traz informações sobre o contexto cultural, o histórico sociopolítico ou a estrutura da sociedade dos países avaliados. Devido a esta ausência de nuances qualitativas, não podemos afirmar com rigor científico que aplicar essas variáveis isoladamente será a "causa" automática e imediata da felicidade de uma nação. A complexidade do bem-estar humano exige que estes dados sejam lidos como fortes tendências, mas não como regras matemáticas infalíveis.
